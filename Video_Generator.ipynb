```python
%%writefile make_audio.py

from gtts import gTTS
import subprocess

text = """Okay robot, lesson one: pasta. Boil water in a pot, add a pinch of salt, then add the pasta once it is bubbling. Stir it so it does not stick, and cook for about nine minutes. Taste a piece to check it is soft, then drain and toss with your sauce.

Lesson two: scrambled eggs. Crack two eggs into a bowl, whisk them with a splash of milk, and pour them into a warm, buttered pan on low heat. Stir gently until just set, then take them off the heat.

Lesson three: grilled cheese. Butter two slices of bread, add a slice of cheese, and cook in a pan on medium low heat until golden and melted.

Lesson four: a simple salad. Chop lettuce, tomato, and cucumber, drizzle with olive oil and lemon, and toss.

Lesson five: a fruit dessert. Slice up whatever fruit you have, and add a spoonful of honey or yogurt on top.

Congratulations robot, you just learned five dishes in one lesson."""

gTTS(text=text, lang="en").save("narration.mp3")

subprocess.run([
    "ffmpeg", "-y",
    "-i", "narration.mp3",
    "-acodec", "pcm_s16le",
    "-ar", "16000",
    "-ac", "1",
    "driven_audio.wav"
], check=True)

print("✅ DONE")
print("Created: narration.mp3")
print("Created: driven_audio.wav")
```

```python
%%writefile make_fried_rice_audio.py

from gtts import gTTS
import subprocess

text = """Okay robot, today we are making simple fried rice. First, cook your rice and let it cool slightly. Heat a pan and add a little oil. Add chopped garlic, carrots, peas, and onions, and stir fry them for a few minutes. Push the vegetables to one side and add one beaten egg. Scramble the egg until cooked. Now add the cooked rice and mix everything together. Add a little soy sauce and stir well. Cook for another two minutes until everything is hot and nicely combined. Finally, add some spring onions on top. Congratulations robot, you just learned how to make fried rice."""

gTTS(text=text, lang="en").save("fried_rice.mp3")

subprocess.run([
    "ffmpeg", "-y",
    "-i", "fried_rice.mp3",
    "-acodec", "pcm_s16le",
    "-ar", "16000",
    "-ac", "1",
    "fried_rice.wav"
], check=True)

print("✅ Fried rice audio created!")
```
